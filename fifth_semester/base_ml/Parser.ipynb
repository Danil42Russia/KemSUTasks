```python
import pandas as pd
```

```python
worlds_list = {
    'first': (1, 'First World Countries Name'),
    'second': (2, 'Second World Countries Name'),
    'third': (3, 'Country Name'),
}

ps = pd.DataFrame()
for name, (index, column_name) in worlds_list.items():
  tables = pd.read_html(f"https://www.burningcompass.com/countries/{name}-world-countries.html", match=column_name)
  assert len(tables) == 1

  table = tables[0]
  assert len(table.columns) >= 10

  table = table[[column_name]]
  table = table.rename(columns={column_name: 'Country'})
  table['Number'] = index
  
  ps = ps.append(table)
```

```python
ps.to_csv('country_economic_rang.csv', sep=',', encoding='utf-8', index=False)
```
