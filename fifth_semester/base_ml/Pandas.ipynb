# Задание 2

Добавить к копии набора данных столбец '**metacritic_rating**' и заполнить его данными из набора данных <a href='https://www.kaggle.com/skateddu/metacritic-all-time-games-stats'>Metacritic all time games stats</a>.

```python
metacritic = pd.read_csv('https://gist.githubusercontent.com/Danil42Russia/0a69528ab4a2133220234187aba83199/raw/2ed49d49c0c6a139b370fc1f4eb2db43a330b030/metacritic_games.csv')
```

Оставляем только нужные столбцы, чтобы при объединении не добавлять лишних данных.

Для объединения оставляем колонки: '**name**' и '**platform**'. Для дополнительного набора данных оставляем '**rating**'.

```python
metacritic = metacritic[['name', 'platform', 'rating']]
```

В данных наборах данных, имена колонок имеют разные имена и разные регистры. Что-бы не усложнять объединение перемеименовываем колонки.

```python
metacritic = metacritic.rename(columns={'name': 'Name', 'platform': 'Platform', 'rating': 'metacritic_rating'})
```

В двух наборах данных, значения для описания платформы имеют разные названия. 

Что-бы при объединение осталось больше данных, меняем названия платформ из набора <a href='https://www.kaggle.com/skateddu/metacritic-all-time-games-stats'>Metacritic all time games stats</a> на <a href='https://www.kaggle.com/gregorut/videogamesales'>Video Game Sales</a>.

```python
values_mapper = {
    'WII':    'Wii',
    'WIIU':   'WiiU',
    'XONE':   'XOne',
    'XBOX':   'XB',
    'VITA':   'PSV',
    'Switch': 'NS',
}

metacritic = metacritic.replace({'Platform': values_mapper})
```

Из-за различия в названиях игр в наборах данных сделаем небольшую хитрость.

Уберём все пробельные и пунктуационные символы их названий игр и добавим нормализованные название в отдельную колонку '**Normalized_Name**'.

```python
import string
def normalized_name(name: str) -> str:
    return name.lower().translate({ord(c): None for c in string.whitespace + string.punctuation})
```

```python
vgsales['Normalized_Name'] = vgsales.apply(lambda row : normalized_name(row['Name']), axis = 1)
```

Названия игр из набора данных <a href='https://www.kaggle.com/skateddu/metacritic-all-time-games-stats'>Metacritic all time games stats</a> для нас не важны, то уберём их.

```python
metacritic['Normalized_Name'] = metacritic.apply(lambda row : normalized_name(row['Name']), axis = 1)
metacritic = metacritic.drop('Name', axis=1)
```

Мерджим значения.

В качестве колонок для пересечения выбрано '**Normalized_Name**' и '**Platform**', потому что одна игра может выйти сразу на нескольких платформах.

```python
data = pd.merge(vgsales, metacritic, how='inner', on=['Normalized_Name', 'Platform'])
data = data.drop('Normalized_Name', axis=1)

data
```

# Задание 5

Вывести жанры игр с количеством игр, которые принадлежат соответствующему жанру в виде '**<жанр> - <количество игр>**' для всех жанров, которые содержат в своем названии не менее 3 различных гласных

Функция для проверки, что текст содержит в своем названии не менее 3 различных гласных.

```python
def is_right_name(word: str) -> bool:
    vowels = ['a', 'e', 'i', 'u', 'y', 'o']
    word = word.lower()

    vowels_in_word = {letter for letter in list(word) if letter in vowels}

    return len(vowels_in_word) >= 3
```

Оставляем только те игры, у которых в названии три уникальных гласных букв.

```python
unique_games = data[data['Name'].apply(lambda x: is_right_name(x))]
```

Группируем игры по жанрам и считаем количество игр в данном жанре. Сортируем по количеству игр в данном жанре от большего к меньшему.

```python
unique_games = unique_games.groupby('Genre').size().reset_index(name='Count').sort_values('Count', ascending=False)
```

Выводим жанры игр с количеством игр

```python
print((unique_games['Genre'].map(str) + ' - ' + unique_games['Count'].map(str)).to_string(index=False))
```
